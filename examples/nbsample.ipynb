## Growth data {#sec-data}

We weighed the puppy[^who] three times. The measurements are in
[@tbl-weights], the subject is pictured in [@fig-puppy], and the analysis
lives in [@sec-math].

[^who]: A border collie, born early 2026.

```
weights = {1: 2.1, 4: 3.4, 8: 5.0}
for week, kg in weights.items(): print(f"week {week}: {kg} kg")
```

## A little analysis {#sec-math}

Average weight is the usual arithmetic mean:

\[
\bar{w} = \frac{1}{n}\sum_{i=1}^{n} w_i
\]

For the data in [-@tbl-weights], \( n = 3 \):

```
avg = sum(weights.values()) / len(weights)
round(avg, 2)
```

So the puppy of [@fig-puppy] averaged \( \bar{w} = 3.5 \) kg over the
period covered by [@tbl-weights] - respectable growth for [@sec-data]'s
eight weeks.
